# Topic Modelling with latent Dirichlet allocation (LDA)
it describes broad task of assigning topics to the unlabeled text documents

# LDA with scikit learn

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
 
df = pd.read_csv("movie_data.csv"  , encoding= "utf-8")


count = CountVectorizer(stop_words= 'english', max_df=0.1, max_features= 5000)

x = count.fit_transform(df['review'].values)

lda = LatentDirichletAllocation(n_components= 10, random_state= 1, learning_method= "batch")

# note n_ components is number of featurem batch fit all data once

x_topics = lda.fit_transform(x)
```

```python
print(lda.components_.shape)
n_top_words = 5

feature_names = count.get_feature_names_out()

for topic_idx,topic in enumerate(lda.components_):
    print(f"Topic {(topic_idx+1)}")
    print(" ".join([feature_names[i] for i in topic.argsort()[:-n_top_words - 1:-1]]))
```

```python
horror = x_topics[:,5].argsort()[::-1]

for iter_idx, movie_idx in enumerate(horror[:3]):
    print(f"\n HOrror movie {(iter_idx +1)}:")
    print(df["review"][movie_idx][:300]," ... ")
```
